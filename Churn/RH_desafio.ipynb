```python
# criar quartis de renda (ou use bins customizados)
df['income_q'] = pd.qcut(df['monthly_income'], q=4, labels=['Q1','Q2','Q3','Q4'])

# taxa de saída por quartil de renda
income_attrition = df.groupby('income_q')['attrition_num'].agg(['mean','count']).rename(columns={'mean':'attrition_rate','count':'n'})
income_attrition['attrition_rate'] = (income_attrition['attrition_rate']*100).round(2)
print(income_attrition)

# mediana de renda entre os que saíram vs ficaram (comparação)
medians = df.groupby('attrition_num')['monthly_income'].median().rename({0:'median_stayed',1:'median_left'})
print('\nMediana MonthlyIncome (0=ficaram,1=sairam):\n', medians)
```

```python
# faixas razoáveis para anos desde última promoção
bins_promo = [-1, 0, 1, 2, 3, 5, 100]
labels_promo = ['0','1','2','3','4-5','6+']
df['years_since_promo_group'] = pd.cut(df['years_since_last_promotion'], bins=bins_promo, labels=labels_promo)

promo_attr = df.groupby('years_since_promo_group')['attrition_num'].agg(['mean','count']).round(3)
promo_attr['attrition_rate_pct'] = (promo_attr['mean']*100).round(2)
print(promo_attr[['attrition_rate_pct','count']])
```

# Principais Insights Sobre a Rotatividade (Attrition)

A taxa geral de rotatividade é de 16,12%, indicando que pouco mais de um em cada seis colaboradores deixou a empresa no período analisado.

Faixa etária: colaboradores entre 26 e 35 anos apresentam a maior incidência de saída, sugerindo que esse grupo pode estar mais propenso a buscar novas oportunidades ou etapas de carreira.

Estado civil: funcionários solteiros demonstram maior taxa de attrition em comparação aos casados ou divorciados.

Gênero: observou-se que homens tiveram uma taxa de saída superior à das mulheres.

Departamento: o setor de Research & Development concentrou o maior número de desligamentos.

Cargos com maior saída: as funções de Laboratory Technician e Sales Executive destacam-se como aquelas com os maiores índices de desligamento.

Nível de cargo: profissionais no Job Level 1 foram os que mais deixaram a organização, possivelmente por estarem em início de carreira ou em posições de menor estabilidade.

Horas extras: a diferença de attrition entre colaboradores que realizam OverTime e os que não realizam é pequena, sugerindo que o impacto dessa variável é limitado no contexto geral.

Work-Life Balance: colaboradores avaliados com nível 3 de equilíbrio entre vida pessoal e profissional apresentaram a maior taxa de saída.

# Atualização

A rotatividade é mais elevada entre colaboradores que estão entre 2 e 3 anos sem promoção (17 - 17.3%), indicando possível sensação de estagnação na carreira nesse período. O grupo com 6+ anos sem promoção também apresenta alta saída (16.3%), reforçando esse efeito ao longo do tempo.
O período crítico para risco de desligamento parece ser entre 2 e 3 anos desde a última promoção, quando o colaborador possivelmente espera progressão na carreira e não a recebe.

O grupo target (idade 26-35 cargo nivel 1 e departamento P&D) apresenta uma taxa de saída muito maior que o restante da empresa, e o principal fator que diferencia esse grupo é a baixa remuneração - sua renda é menos da metade da dos outros funcionarios.
Como o tempo desde a última promoção é igual nos dois grupos, a questão salarial é, com altíssima probabilidade, o fator mais determinante da alta saída do grupo.

Há diferença salarial significativa entre os grupos (Mann-Whitney U = 100620.5, p < 0.001). O grupo com maior taxa de desligamento apresenta salários consistentemente menores, reforçando a remuneração como um fator crítico associado à rotatividade.


#Atualização 02

A satisfação no trabalho não diferencia o grupo de alto risco do restante da empresa: ambos têm mediana 3. Contudo, dentro do grupo, os colaboradores com satisfação baixa têm risco até 2x maior de desligamento. Isso indica que a causa estrutural da rotatividade não é insatisfação, mas sim fatores externos como remuneração e progressão.

O fator gestor é crítico.
O turnover cai drasticamente sob líderes com 2–3 anos de casa, e dispara entre gestores novos (0–1 ano) e muitos antigos (6+ anos).

Os testes qui-quadrado mostram que nem a satisfação no trabalho (p = 0.15) nem o desempenho (p = 0.16) têm associação estatística significativa com o turnover.
Isso indica que a saída dos funcionários não é seletiva e não é dirigida por baixa satisfação nem por baixa performance.

Pessoas com 2 anos sem promoção têm uma das maiores taxas de saída, considerando que:
pós 2 anos, a expectativa de crescimento aumenta — e a falta dele gera saída.

Funcionários com gestores muito novos (0–1 ano de supervisão) apresentam as maiores taxas de saída:
0 anos: 36.6%
1 ano: 33.3%

O teste Qui-Quadrado mostrou:
p = 0.1507 → Sem evidência estatística de que níveis de satisfação influenciam a saída.

Funcionários de performance alta e média saem em proporções semelhantes.

A empresa não perde funcionários porque eles estão insatisfeitos ou têm desempenho ruim — ela perde porque não os desenvolve, não os remunera adequadamente e não garante liderança estável.

O foco central é capacitar líderes para reduzir turnover e aumentar a estabilidade de suas equipes no período crítico dos primeiros 12 meses. O programa combina treinamento, acompanhamento e ferramentas práticas que aceleram a maturidade da liderança e aumentam a eficácia na gestão de pessoas.
